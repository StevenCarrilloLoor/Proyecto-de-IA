# Guía Rápida - Sistema de Análisis de Baloncesto

Este notebook muestra cómo usar el sistema de análisis de baloncesto para procesar videos y obtener estadísticas.

## 1. Importar Librerías

```python
import sys
sys.path.insert(0, '../src')

from main import BasketballAnalyzer
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

## 2. Inicializar el Analizador

```python
# Crear instancia del analizador
analyzer = BasketballAnalyzer(config_path='../config/config.yaml')

print("✓ Analizador inicializado correctamente")
```

## 3. Procesar un Video

```python
# Ruta al video (reemplazar con tu video)
video_path = "../data/raw/sample_game.mp4"
output_path = "../data/processed/analyzed_game.mp4"

# Procesar video
results = analyzer.process_video(
    video_path=video_path,
    output_path=output_path,
    generate_heatmap=True,
    classify_plays=True,
    show_progress=True
)
```

## 4. Visualizar Estadísticas

```python
# Mostrar reporte de estadísticas
print(analyzer.stats_calc.format_report(results['statistics']))
```

## 5. Visualizar Heatmaps

```python
# Visualizar heatmap de jugadores
if 'heatmaps' in results:
    fig, axes = plt.subplots(1, 2, figsize=(15, 6))
    
    # Heatmap de jugadores
    axes[0].imshow(results['heatmaps']['players'], cmap='hot')
    axes[0].set_title('Heatmap de Jugadores')
    axes[0].axis('off')
    
    # Heatmap del balón
    axes[1].imshow(results['heatmaps']['ball'], cmap='hot')
    axes[1].set_title('Heatmap del Balón')
    axes[1].axis('off')
    
    plt.tight_layout()
    plt.show()
```

## 6. Analizar Jugadas Clasificadas

```python
# Distribución de tipos de jugadas
if 'play_classifications' in results and results['play_classifications']:
    from collections import Counter
    
    play_types = [p['type'] for p in results['play_classifications']]
    distribution = Counter(play_types)
    
    plt.figure(figsize=(10, 6))
    plt.bar(distribution.keys(), distribution.values())
    plt.xlabel('Tipo de Jugada')
    plt.ylabel('Frecuencia')
    plt.title('Distribución de Jugadas Detectadas')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

## 7. Guardar Resultados

```python
# Guardar resultados completos
output_dir = "../results/analysis_output"
analyzer.save_results(results, output_dir)

print(f"✓ Resultados guardados en: {output_dir}")
```
